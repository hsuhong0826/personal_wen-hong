```python
#時間1
x1 = int(input())
x2 = int(input())

#時間2
y1 = int(input())
y2 = int(input())

x = 0
y = 0
time = 0
money = 0

#計算時間
x = x1 * 60 + x2
y = y1 * 60 + y2

#計算時間差
time = y - x
    
if(time < 30):
    print(0)
elif(time >= 30 and time <= 120):
    money = (time // 30) * 30
    print(money)
elif(time > 120 and time < 240):
    money = 120 + ((time - 120) // 30) * 40
    print(money)
else:
    money = 120 + 160 + ((time - 240) // 30) * 60
    print(money)
```

```python
n = int(input())
sum = 0

for i in range(1, n + 1):  #從 1 ~ n
    if i % 3 == 0:
        sum += i

print(sum)
```

```python
def gcd(n, m):
    if n == m or n % m == 0:  # n跟m一樣 or m是n的因數
        return m
    else:
        return gcd(m, n % m)  #輾轉相除

n = int(input())
m = int(input())

print(gcd(n, m))
```

```python
def f(n):
    if n == 0 or n == 1:
        return n + 1
    else:
        return f(n - 1) + f(n // 2)

n = int(input())
print(f(n))
```

```python
n = list(input())  #轉型態-->['1', '2', '3', '4', '3', '2', '1']

i = 0  #頭
j = len(n)-1  #尾

text = '是'

while(i < len(n)):
    if(n[i] != n[j]):  #如果不同就不是迴文
        text = '否'
        break
    else:
        i = i+1  #往下一個跑
        j = j-1  #往前一個跑
        
print(text)
```
